# Setup — point this notebook at the data

The omics files are **not** in the GitHub repository (some are large, e.g. the RNA matrix is
~70 MB). Download them from the **course page** into a single folder, then set `DATA_DIR`.

- **Local / Jupyter:** set `DATA_DIR` to that folder (e.g. `"data/multiomics"`).
- **Google Colab:** either mount Google Drive and point `DATA_DIR` there, or run the upload
  helper in the commented block.

**Important — how the two files line up:** the data matrix and the annotation table have the
**same number of rows in the same order**. There is *no* shared ID column, so you join them
**by position** (row *i* of the matrix corresponds to row *i* of the annotation), not with a
`merge` on a key.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ----- choose the omics layer and where the files live -----
LAYER = "RNA"                 # RNA | Proteomics | Metabolomics | Cytokines | Nasal | Gut | Clinical_labs
DATA_DIR = Path("data/multiomics")   # <-- folder where you downloaded the course-page files

# On Google Colab you can instead upload the two files for your chosen layer:
# from google.colab import files; files.upload()
# DATA_DIR = Path(".")

data_path = DATA_DIR / f"{LAYER}_df_Data.csv"
anno_path = DATA_DIR / f"{LAYER}_annotation_colData.csv"

omics = pd.read_csv(data_path)                 # rows = samples, columns = features
anno  = pd.read_csv(anno_path)                 # rows = same samples, same order

assert len(omics) == len(anno), "Matrix and annotation must have the same number of rows"

# Positional join: attach annotation columns alongside the feature matrix
anno = anno.reset_index(drop=True)
omics = omics.reset_index(drop=True)
# Dates are formatted YY-MM-DD HH:MM:SS; a few are genuinely missing (-> NaT)
anno["Date"] = pd.to_datetime(anno["Date"], format="%y-%m-%d %H:%M:%S", errors="coerce")
print("Missing dates (NaT):", anno["Date"].isna().sum())

print(f"{LAYER}: {omics.shape[0]} samples x {omics.shape[1]} features")
print("Groups (IRIS):", anno["IRIS"].value_counts().to_dict())
print("Subjects:", anno["SubjectID"].nunique())
anno.head()
```
